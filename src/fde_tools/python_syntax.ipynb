Python structures:
1. Lists - square brackets - ordered
2. Tuples - round brackets - ordered, immutable.
3. Dictionary - curly brackets - Key value pairs, unordered
4. Set - curly brackets - unordered, mutable, UNIQUE values

```python
# List
my_colors = ['red', 'blue', 'green', 'yellow', 'pink', 'orange', 'white']


print(f'my_colors has {len(my_colors)} elements')

print(f'the first element of my_colors is {my_colors[0]}.')
print(f'the last element of my_colors is {my_colors[-1]}.')

for color in my_colors:
    print(color)


for index, color in enumerate(my_colors):
    print(f'the {index} element of my_colors is {color}')


# slicing x:y from x element to before y element
last_three_colors = my_colors[1:2]


for index, color in enumerate(last_three_colors):
    print(f'the {index} element of last_three_colors is {color}')

```

```python
# Reversing a list
my_list = [1, 2, 3, 4]
print('list is', my_list)

reversed_my_list = my_list[::-1]

print('reverse list is: ', reversed_my_list)

# .reverse() reverses the list in place
my_list.reverse()
print('reversed list is', my_list)

# reversed() returns a lazy iterator
for x in reversed(my_list):
    print(x)
```

```python
# Tuples

my_tuple = ('red', 'green', 'blue')

print (my_tuple)

print(f'The length of my tuple is {len(my_tuple)}')

print(my_tuple[0])

for my_tup in my_tuple:
    print(my_tup)

new_tup = my_tuple[1:3]
print(new_tup)

for index, color in enumerate(my_tuple):
    print(f'the {index} element of my_tuple is {color}')
```

```python
# Dictionary

my_dictionary = {'red': '#ff0000', 'green': '#00ff00', 'blue': '#0000ff'}

print(my_dictionary)

print(len(my_dictionary))

for key in my_dictionary:
    print(key)

for key, value in my_dictionary.items():
    print(f'Key is {key}. Value is {value}')
```

```python
# Set

my_set = {'user1', 'user2', 'user3'}

if "user1" in my_set:  # Blazingly fast O(1) lookup
    print("user1 Access granted")
else:
    print("user1 Access denied")

if "user4" in my_set:  # Blazingly fast O(1) lookup
    print("user4 Access granted")
else:
    print("user4 Access denied")
```

For loop with range as Python doesn't have a typical for loop. for i in range(1,5): it will have value 1 to 4.

```python
for i in range(1, 5):
    print('The number is', i)
```

lambda function is a single expression function defined with equal to sign instead of writing def function_name: 

```python
# Regular function
def square(x):
    return x * x

print('regular funciton', square(3))

# Equivalent lambda
lambda_square = lambda x: x * x

print('lambda funciton', lambda_square(5))  # 25
```

```python
list1 = [4,5,6,7]
my_list = list(map(lambda x: x//2, list1))
print(my_list)
```
