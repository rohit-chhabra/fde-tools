RAG Advanced Techniques
1. Chunking R&D: experiment with chunking strategy. What should be chunk size? How much should be overlap? 
2. Encoder R&D: select the best Encoder model based on a test set. Encoder creates vectors.
3. Improve Prompts: experiment with the base prompts, add relevant content, the current date, and history.
4. Document Preprocessing: use an LLM to rewrite document before encoding. Ask LLM to rewrite document which is suitable for vectorization and vector matching. Ask LLM to make chunks or use semantic chunking.
5. Query Preprocessing / rewriting: use an LLM to convert the user's question to a RAG query.
6. Query expansion: use an LLM to turn the question into multiple RAG queries.
7. Re-ranking: use an LLM to reorder and sub-select from RAG results.
8. Hierarchical: use an LLM to summarize at multiple levels to answer hierarchical questions.
9. Graph RAG: Documents have relationships. Chunks are related to other chunks. eg. other chunks in the same documents.
10. Agentic RAG: Could be a Vector based RAG or let Agent decide if they need to use a tool (could be a vector lookup or database lookup or internet search or call an API)
